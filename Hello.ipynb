2. Complarison Operatiors


```python
x = 10
y = 5

print(x == y)  # is x equal to y?
print(x != y)  # is x not equal to y?
print(x < y)   # is x less than y?
print(x > 7)   # is x grater than y?
```

3. Logical Operators


```python
age = 25

print(age > 18 and age < 60)   # both must be true
print(age < 10 or age > 18)    # any one true is enough
print(not(age > 18))           # reverses true or false


```

Student Grade Manager


```python
students = {}

def calculate_grade(avg):
    if avg >= 90:
        return "A"
    elif avg >= 80:
        return "B"
    elif avg >= 70:
        return "C"
    elif avg >= 60:
        return "D"
    else:
        return "F"

while True:
    print("\n1. Add Student")
    print("2. View Students")
    print("3. Exit")

    choice = input("Enter choice: ")

    if choice == "1":
        name = input("Student Name: ")
        marks = list(map(int, input("Enter marks (3 subjects): ").split()))
        
        total = sum(marks)
        avg = total / len(marks)
        grade = calculate_grade(avg)

        students[name] = {
            "marks": marks,
            "average": avg,
            "grade": grade
        }

        print("Student added successfully!")

    elif choice == "2":
        for name, data in students.items():
            print(f"\nName: {name}")
            print(f"Marks: {data['marks']}")
            print(f"Average: {data['average']:.2f}")
            print(f"Grade: {data['grade']}")

    elif choice == "3":
        break

    else:
        print("Invalid choice!")
```

```python
students = {
    "Mohammed": [],
    "Rizwan": [],
    "Syeda": [],
    "Keerthi": [],
    "Karthi": []
}

def add_grade(student, grade):
    if student in students:
        students[student].append(grade)
        print(f"Grade {grade} added for {student}")
    else:
        print("Student not found")

def show_grades():
    print("\n--- Student Grades ---")
    for student, grades in students.items():
        if grades:
            average = sum(grades) / len(grades)
            print(f"{student}: Grades = {grades}, Average = {average:.2f}")
        else:
            print(f"{student}: No grades available")

# Example usage
add_grade("Mohammed", 85)
add_grade("Rizwan", 90)
add_grade("Syeda", 78)
add_grade("Keerthi", 88)
add_grade("Karthi", 95)

show_grades()
```
